##### Avaliação dos atributos e classificação
      - Verificando quais colunas são numéricas e categóricas.:

```python
# Verificando quais colunas são numéricas e categóricas.

numeric_features = [feature for feature in df.columns
                    if df[feature].dtype != 'O']

categorical_features = [feature for feature in df.columns
                        if df[feature].dtype == 'O']

# Print das colunas
print(f"Temos {len(numeric_features)} numeric_features: {numeric_features}")
print(f"Temos {len(categorical_features)} categorical_features: {categorical_features}")
```

##### Análise Univariada

```python
warnings.filterwarnings('ignore')
```

```python
#Variáveis Numéricas
plt.figure(figsize=(15,15))
plt.suptitle("Análise univariada de atributos numéricos", fontsize=20, fontweight='bold',alpha=0.8, y=1)

for i in range(len(numeric_features)):
  plt.subplot(5,3,i+1)
  sns.kdeplot(x=df[numeric_features[i]],shade=True,color='b')
  plt.xlabel(numeric_features[i])
  plt.tight_layout()
```
