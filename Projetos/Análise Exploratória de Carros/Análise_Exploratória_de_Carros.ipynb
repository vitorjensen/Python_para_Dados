##### Análise Exploratória

```python
# Print de dados únicos em colunas específicas
print("Categoria em 'seller_type:'",end=" ")
print(df['seller_type'].unique())
print("\n")

print("Categorias em 'fuel_type:'", end=" ")
print(df['fuel_type'].unique())
print("\n")

print("Categorias em 'transmission_type:'", end=" ")
print(df['transmission_type'].unique())
print("\n")

print("Categorias em 'seats:'", end=" ")
print(df['seats'].unique())
print("\n")
```

##### Avaliando Possíveis inconsistências de Dados

```python
# Checagem das colunas e possíveis erros.
df[df['seats'] == 0]
```

```python
# Reescrevendo os valores de 0 presentes nas linhas da coluna "seats" para 5
df.loc[df['seats'] == 0, 'seats'] = 5

# Ou se desejar remover as linhas contendo zero, use:
# df.drop(df[df['seats'] == 0].index, inplace=True)
```
