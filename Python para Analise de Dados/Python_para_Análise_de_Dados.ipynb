### Binding ( Vinculação ) em Python.

- A "atribuição" ou também Binding (Vinculação), acontece quando atribuímos o valor inicial de uma variável a outra. Por exemplo:

```python
a = [1, 2, 3] # "a" é uma variável vinculada
```

### Instâncias, Classes e a função: isinstance()
- **Instância:** Quando um objeto é criado e tipado com o "molde" de uma Classe, representando algo real.
- **Classe:** Representa "moldes" a serem atribuídos a objetos.
- **Função: isinstance():** Serve função para verificar qual o tipo de Classe (molde) o objeto foi instanciado. Retorna booleano, True ou Falso.

```python
a = 5
type(a)
isinstance(a, int)
```

- A função "isinstance()", pode receber uma tupla para fazer a verificação.

```python
c = 2; b = 2.5
isinstance(c, (int, float)) # Faz uma verificação da variável "OR" da variável C entre os dois tipo.
isinstance(b, (int, float))
```

### Duck Typing:

- Não se preocupa com a classe que o objeto foi instanciado, apenas se tiver o método desejado a executar.

```python
def isiterabel(obj):
  try:
    iter(obj)
    return True
  except TypeError:
    return False
```

```python
isiterabel("a string")
isiterabel([1, 2, 3])
isiterabel(5)
```

### String: Templating

- A **formatação** de string é um tema importante. Os objetos **string** possuem um método **format**, que pode ser usado para substituir argumentos formatados na string, produzindo uma nova string

```python
template = "{0:.2f} {1:s} are worth US${2:d}"
```

{0:.2f} -> Significa formatar o primeiro argumento como um número de ponto flutuante com duas casas decimais.

{1:s} -> Significa formataro segundo argumento como **String.**

{2:d} -> Significa formatar o terceiro argumento como um inteiro exato.

```python
template.format(88.46, "Argentine Pesos", 1)
```

### F-Strings (Formatted String literals):
- Para criar, insira o caractere 'f' imediatamente antes de uma string literal. Use chaves para fazer a substituição de seu valor na string formatada.

```python
amount = 10
rate = 88.46
currency = "Pesos"
```

```python
result = f"{amount} {currency} is worth US${amount / rate}"
print(result)
```
