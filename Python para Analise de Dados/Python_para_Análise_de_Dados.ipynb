### Binding ( Vinculação ) em Python.

- A "atribuição" ou também Binding (Vinculação), acontece quando atribuímos o valor inicial de uma variável a outra. Por exemplo:

```python
a = [1, 2, 3] # "a" é uma variável vinculada
```

```python
b = a
print(b)
```

- Quando passamos objetos como argumentos para uma função, novas variáveis locais são criadas referenciando os objetos originais.

- Uma variável de mesmo nome dentro de uma função (escopo filho) armazenando um objeto novo, não sobrescreve uma variável fora do escopo da função (escopo pai).

- **Argumento mutável:** Por exemplo: Uma Lista em Python. Acontece quando definimos um valor ao argumento em sua criação. Pode ser alterado.

### Instâncias, Classes e a função: isinstance()
- **Instância:** Quando um objeto é criado e tipado com o "molde" de uma Classe, representando algo real.
- **Classe:** Representa "moldes" a serem atribuídos a objetos.
- **Função: isinstance():** Serve função para verificar qual o tipo de Classe (molde) o objeto foi instanciado. Retorna booleano, True ou Falso.

```python
a = 5
type(a)
isinstance(a, int)
```

- A função "isinstance()", pode receber uma tupla para fazer a verificação.

```python
c = 2; b = 2.5
isinstance(c, (int, float)) # Faz uma verificação da variável "OR" da variável C entre os dois tipo.
isinstance(b, (int, float))
```

### Duck Typing:

- Não se preocupa com a classe que o objeto foi instanciado, apenas se tiver o método desejado a executar.

```python
def isiterabel(obj):
  try:
    iter(obj)
    return True
  except TypeError:
    return False
```

```python
isiterabel("a string")
isiterabel([1, 2, 3])
isiterabel(5)
```
